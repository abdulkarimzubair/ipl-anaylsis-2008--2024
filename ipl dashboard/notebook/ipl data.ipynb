```python
batting = pd.read_csv("all_season_batting_card.csv")
bowling =pd.read_csv("all_season_bowling_card.csv")
summary = pd.read_csv("all_season_summary.csv")
details = pd.read_csv("all_season_details.csv")
```

```python
batting_merged_df = pd.concat([batting, batting_2024], ignore_index = True)
bowling_merged_df = pd.concat([bowling, bowling_2024], ignore_index = True)
summary_merged_df = pd.concat([summary, summary_2024], ignore_index = True)
details_merged_df = pd.concat([details, details_2024], ignore_index = True)
```

```python
print(batting_merged_df.columns)
print(bowling_merged_df.columns)
print(summary_merged_df.columns)
print(details_merged_df.columns)
```

```python
batting_merged_df["season"] = batting_merged_df["season"].ffill()
bowling_merged_df["season"] = bowling_merged_df["season"].ffill()
summary_merged_df["season"] = summary_merged_df["season"].ffill()
details_merged_df["season"] = details_merged_df["season"].ffill()
```

```python
pattern =r"\bwon\b"
match_played = summary_merged_df["result"].str.contains(pattern,flags = re.I,na = False)
summary_merged_df.loc[match_played,"match_status"] = "match completed"
summary_merged_df
```

```python
pattern =r"\babandoned\b"


match_abandoned = summary_merged_df["result"].str.contains(pattern,flags = re.I,na = False)
summary_merged_df.loc[match_abandoned, "winner"] = "draw"
summary_merged_df.loc[match_abandoned, "match_status"] = "draw"
```

```python
pattern =r"^\bno\b"
match_draw = summary_merged_df["result"].str.contains(pattern,flags = re.I,na = False)
summary_merged_df.loc[match_draw, "winner"] = "draw"
summary_merged_df.loc[match_draw, "match_status"] = "draw"
```

```python
pattern =r"^\bstarts\b"
match_notplayed_yet = summary_merged_df["result"].str.contains(pattern,flags = re.I,na = False)
summary_merged_df.loc[match_notplayed_yet, "winner"] = "soon"
summary_merged_df.loc[match_notplayed_yet, "match_status"] = "not played yet"
```

```python
summary_merged_df["name"].str.split("v", n = 1, expand = True)
summary_merged_df["home_team_name"] = summary_merged_df["name"].str.split(" v ", n = 1, expand = True)[0]
summary_merged_df["away_team_name"] = summary_merged_df["name"].str.split(" v ", n = 1, expand = True)[1]
```

```python
summary_merged_df["match"] = summary_merged_df["description"].str.split(",", n = 1, expand = True)[0].str.rsplit( "(", expand = True)[0]
```

```python
summary_merged_df
```

```python
import re

pattern = r"\b(^\d\w+\s+Match)\b"

summary_merged_df["stage"] = summary_merged_df["match"].str.contains(pattern, flags=re.I, na = False).map({True : "league stage", False: "playoffs"})
```

```python
summary_merged_df
```

```python
def balls(x):
    if pd.isna(x):
        return 0
    parts = str(x).split(".")
    overs = int(parts[0])
    ball_left = int(parts[1]) if len(parts) > 1 else 0
    return overs * 6 + ball_left
```

```python
summary_merged_df["home_balls"] = summary_merged_df["home_overs"].apply(balls)
summary_merged_df["away_balls"] = summary_merged_df["away_overs"].apply(balls)
```

```python
display(summary_merged_df["home_team_name"].unique())
```

```python
summary_merged_df["match"] = summary_merged_df["match"].str.strip()
```

```python
qf = ["2nd Qualifying Match", "2nd Qualifying Final"]

summary_merged_df.loc[summary_merged_df["match"].isin(qf), "match"] = "Qualifier 2"
        
```

```python
home_team_table = summary_merged_df[["season","home_team","home_team_name","winner","match","home_runs","away_runs","home_balls","away_balls","stage","match_status"]] \
.rename(columns={"home_team": "team",
                 "home_team_name": "team_full_name",
                 "home_runs":"runs_scored",
                 "away_runs": "runs_conceded",
                 "home_balls": "balls_played",
                 "away_balls": "balls_bowled"})
away_team_table = summary_merged_df[["season","away_team","away_team_name","winner","match","home_runs","away_runs","home_balls","away_balls","stage","match_status"]]\
.rename(columns={"away_team": "team",
                 "away_team_name": "team_full_name",
                 "away_runs":"runs_scored",
                 "home_runs": "runs_conceded",
                 "away_balls": "balls_played",
                 "home_balls": "balls_bowled"})
home_team_table
```

```python
points_table = pd.concat([home_team_table,away_team_table],ignore_index = True)
points_table
```

```python
points_table["played"] = (points_table["match_status"] != "not played yet").astype(int)
points_table["win"] = (points_table["team"] == points_table["winner"]).astype(int)
points_table["nr"] = (points_table["winner"] == "draw").astype(int)
points_table["lost"] =  points_table["played"] - points_table["win"] - points_table["nr"]
points_table["point"] = abs(points_table["win"] * 2 - points_table["nr"])
points_table["final_played"] = points_table["match"] == "Final"
points_table["champions"] = (points_table["final_played"] == True) & (points_table["team"] == points_table["winner"])
points_table["runner_up"] = (points_table["final_played"] == True) & (points_table["team"] != points_table["winner"])
is_q2_loser = (points_table["match"] == "Qualifier 2") & (points_table["team"] != points_table["winner"]) 
is_third_place_winner = (points_table["match"] == "3rd Place Play-off") & (points_table["team"] == points_table["winner"]) 
semi_first = (points_table["match"] == "3rd Place Play-off")
points_table["runner_up_2nd"] = is_q2_loser | is_third_place_winner

                                
```

```python
no_score = (points_table["nr"] == 1)
points_table.loc[no_score,"runs_scored"] = 0
points_table.loc[no_score,"runs_conceded"] = 0
points_table.loc[no_score,"balls_played"] = 0
points_table.loc[no_score,"balls_bowled"] = 0
```

```python
league_table = points_table[(points_table["stage"] == "league stage") & (points_table["match_status"] != "not played yet")]
```

```python
cols = ["runs_scored","runs_conceded","balls_played","balls_bowled","played","win","nr","lost","point"]
bol_cols = ["champions","runner_up","runner_up_2nd"]

flags = (points_table
         .groupby(["season", "team","team_full_name"])[bol_cols]
         .any()
         .reset_index())

team_ranking = (league_table.groupby(["season","team",])[cols]
                            .sum()
                            .reset_index()
                            .merge(flags, on=["season", "team"], how="left"))
team_ranking 

```

```python
team_ranking["nrr"] = ((team_ranking["runs_scored"] * 6 / team_ranking["balls_played"])  
                      - (team_ranking["runs_conceded"] * 6 / team_ranking["balls_bowled"]))
```

```python
batsman_vs_bowler = details_merged_df[['season','match_id','match_name','home_team',
       'away_team','current_innings','innings_id', 'over', 'ball', 'runs',
       'shortText', 'isBoundary', 'isWide', 'isNoball', 'batsman1_id',
       'batsman1_name', 'batsman1_runs', 'batsman1_balls', 'bowler1_id',
       'bowler1_name', 'bowler1_overs', 'bowler1_maidens', 'bowler1_runs',
       'bowler1_wkts',]]
```

```python
batsman_vs_bowler
```

```python
bat_vs_bowl = (
    batsman_vs_bowler.groupby(["season","match_id","batsman1_id", "bowler1_id"])
    .agg(runs=("runs", "sum"),
         balls=("bowler1_id", "count"))
    .reset_index()
)
bat_vs_bowl
```

```python
bat_vs_bowl["strike_rate"] = round((bat_vs_bowl["runs"] / bat_vs_bowl["balls"])* 100,2 )
bat_vs_bowl["economy"] = round((bat_vs_bowl["runs"] / bat_vs_bowl["balls"])* 6,2 )

bat_vs_bowl
```

```python
bat_vs_bowl.to_csv('bat_vs_bowl.csv', index=False)
```
